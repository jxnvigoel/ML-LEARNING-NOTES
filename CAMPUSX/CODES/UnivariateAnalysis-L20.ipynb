```python
import pandas as pd 
import seaborn as sns
df=pd.read_csv("CAMPUSX/DATASETS/train.csv")
```

```python
df.head()
```

1.CATEGORICAL DATA:

a.Countplot

```python
sns.countplot(x=df["Sex"])
```

b.PieChart

```python
df['Sex'].value_counts().plot(kind='pie',autopct='%.2f')
```

a.Histogram

```python
import matplotlib.pyplot as plt
plt.hist(df["Age"],bins=5)
```

b.Distplot

```python
sns.distplot(df["Age"])
```

c.Boxplot

```python
sns.boxplot(df["Age"])
```
