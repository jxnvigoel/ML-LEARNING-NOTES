```python
import pandas as pd
import seaborn as sns
```

seaborn already has datasets on which we can perform analysis so we will use that datasets also


```python
tips= sns.load_dataset('tips')
tips.head()
```

```python
flight=sns.load_dataset("flights")
flight.head()
```

```python
iris=sns.load_dataset('iris')
iris.head()
```

```python
titanics=pd.read_csv("CAMPUSX/DATASETS/train.csv")
titanics.head()
```

1. SCATTER PLOT (Numerical - Numerical)

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'],style=tips['smoker'],size=tips['size'])
```

2.BAR PLOT (Numerical-catergorical)

```python
sns.barplot(x=titanics['Pclass'],y=titanics['Age'])
```

```python
sns.barplot(x=titanics['Pclass'],y=titanics['Age'],hue=titanics['Sex'])
```

```python
sns.scatterplot(y=titanics['Age'],x=titanics['Pclass'])
```

3. BOXPLOT (Numerical-categorical)

```python
sns.boxplot(x=titanics['Sex'],y=titanics["Age"],hue=titanics["Survived"])
```

```python
sns.boxplot(x=tips['sex'],y=tips['tip'],hue=tips['smoker'])
```

4.DISTPLOT(Num-Categorical)

```python
sns.distplot(titanics[titanics['Survived']==0]['Age'],hist=False)
sns.distplot(titanics[titanics['Survived']==1]['Age'],hist=False)

```

5. HEATMAP(Catergorical-categorical)

```python
# pd.crosstab(titanics["Pclass"],titanics["Survived"])
sns.heatmap(pd.crosstab(titanics["Pclass"],titanics["Survived"]))
```

```python
(titanics.groupby('Pclass')['Survived'].mean()*100).plot(kind='bar')
```

6.CLUSTER MAPING(categ-categ)

```python
%pip install scipy
```

```python
sns.clustermap(pd.crosstab(titanics["Pclass"],titanics["Survived"]))
```

7.PAIRPLOT(used to compare many columns with each other )

```python
sns.pairplot(iris,hue="species")
```

8.LINEPLOT(usually used when u have timebased data on x axis)

```python
new=flight.groupby('year')['passengers'].sum().reset_index()
new
```

```python
sns.lineplot(x=new['year'],y=new['passengers'])
```

```python
sns.clustermap(flight.pivot_table(values='passengers',index='month',columns='year'))
```
